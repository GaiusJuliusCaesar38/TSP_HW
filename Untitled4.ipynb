```python
import os
import re
import math
import time
import random
import itertools
from dataclasses import dataclass
from typing import Dict, Tuple, List, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
@dataclass
class CVRPInstance:
    name: str
    dimension: int
    capacity: int
    coords: Dict[int, Tuple[float, float]]
    demands: Dict[int, int]
    depot: int
    distance_matrix: np.ndarray


@dataclass
class ACOResult:
    best_cost: float
    best_routes: List[List[int]]
    history: List[float]
    elapsed_time: float
    iterations_done: int


def euclidean_distance(x1, y1, x2, y2):
    return math.hypot(x1 - x2, y1 - y2)


def build_distance_matrix(coords: Dict[int, Tuple[float, float]]) -> np.ndarray:
    nodes = sorted(coords.keys())
    idx = {node: i for i, node in enumerate(nodes)}
    n = len(nodes)
    dist = np.zeros((n, n), dtype=float)

    for a in nodes:
        for b in nodes:
            i = idx[a]
            j = idx[b]
            x1, y1 = coords[a]
            x2, y2 = coords[b]
            dist[i, j] = euclidean_distance(x1, y1, x2, y2)

    return dist


def node_to_index_map(instance: CVRPInstance) -> Dict[int, int]:
    return {node: i for i, node in enumerate(sorted(instance.coords.keys()))}


def route_cost(route: List[int], instance: CVRPInstance) -> float:
    n2i = node_to_index_map(instance)
    total = 0.0
    for a, b in zip(route[:-1], route[1:]):
        total += instance.distance_matrix[n2i[a], n2i[b]]
    return total


def solution_cost(routes: List[List[int]], instance: CVRPInstance) -> float:
    return sum(route_cost(route, instance) for route in routes)

def two_opt_route(route, instance, max_loops=3, max_checks=100):
    if len(route) <= 4:
        return route

    best_route = route[:]
    best_cost = route_cost(best_route, instance)
    loops = 0

    while loops < max_loops:
        loops += 1
        improved = False
        checks = 0

        for i in range(1, len(best_route) - 2):
            for j in range(i + 1, len(best_route) - 1):
                if j - i == 1:
                    continue

                checks += 1
                if checks > max_checks:
                    return best_route

                new_route = best_route[:i] + best_route[i:j][::-1] + best_route[j:]
                new_cost = route_cost(new_route, instance)

                if new_cost < best_cost:
                    best_route = new_route
                    best_cost = new_cost
                    improved = True
                    break
            if improved:
                break

        if not improved:
            break

    return best_route

def improve_routes_2opt(routes, instance):
    return [two_opt_route(route, instance, max_loops=1, max_checks=100) for route in routes]

def route_load(route, instance):
    return sum(instance.demands[node] for node in route if node != instance.depot)

def relocate_one_customer(routes, instance):
    best_routes = [r[:] for r in routes]
    best_cost = solution_cost(best_routes, instance)

    loads = [route_load(route, instance) for route in best_routes]

    for from_idx in range(len(best_routes)):
        from_route = best_routes[from_idx]
        if len(from_route) <= 3:
            continue

        for pos in range(1, len(from_route) - 1):
            customer = from_route[pos]
            demand = instance.demands[customer]

            for to_idx in range(len(best_routes)):
                if from_idx == to_idx:
                    continue

                to_route = best_routes[to_idx]

                if loads[to_idx] + demand > instance.capacity:
                    continue

                new_from = from_route[:pos] + from_route[pos + 1:]

                for insert_pos in range(1, len(to_route)):
                    new_to = to_route[:insert_pos] + [customer] + to_route[insert_pos:]

                    candidate_routes = [r[:] for r in best_routes]
                    candidate_routes[from_idx] = new_from
                    candidate_routes[to_idx] = new_to

                    if len(candidate_routes[from_idx]) < 2:
                        continue

                    candidate_cost = solution_cost(candidate_routes, instance)

                    if candidate_cost + 1e-9 < best_cost:
                        return candidate_routes, True

    return best_routes, False

def improve_routes_relocate(routes, instance, max_loops=10):
    best_routes = [r[:] for r in routes]
    loops = 0

    while loops < max_loops:
        loops += 1
        best_routes, improved = relocate_one_customer(best_routes, instance)
        if not improved:
            break

    return best_routes

def swap_two_customers(routes, instance):
    best_routes = [r[:] for r in routes]
    best_cost = solution_cost(best_routes, instance)

    loads = [route_load(route, instance) for route in best_routes]

    for r1_idx in range(len(best_routes)):
        route1 = best_routes[r1_idx]
        if len(route1) <= 2:
            continue

        for r2_idx in range(r1_idx + 1, len(best_routes)):
            route2 = best_routes[r2_idx]
            if len(route2) <= 2:
                continue

            for p1 in range(1, len(route1) - 1):
                c1 = route1[p1]
                d1 = instance.demands[c1]

                for p2 in range(1, len(route2) - 1):
                    c2 = route2[p2]
                    d2 = instance.demands[c2]

                    new_load1 = loads[r1_idx] - d1 + d2
                    new_load2 = loads[r2_idx] - d2 + d1

                    if new_load1 > instance.capacity or new_load2 > instance.capacity:
                        continue

                    new_route1 = route1[:]
                    new_route2 = route2[:]

                    new_route1[p1] = c2
                    new_route2[p2] = c1

                    candidate_routes = [r[:] for r in best_routes]
                    candidate_routes[r1_idx] = new_route1
                    candidate_routes[r2_idx] = new_route2

                    candidate_cost = solution_cost(candidate_routes, instance)

                    if candidate_cost + 1e-9 < best_cost:
                        return candidate_routes, True

    return best_routes, False


def improve_routes_swap(routes, instance, max_loops=10):
    best_routes = [r[:] for r in routes]
    loops = 0

    while loops < max_loops:
        loops += 1
        best_routes, improved = swap_two_customers(best_routes, instance)
        if not improved:
            break

    return best_routes

def or_opt_route(route, instance, block_sizes=(1, 2), max_loops=1):
    if len(route) <= 4:
        return route

    best_route = route[:]
    best_cost = route_cost(best_route, instance)
    loops = 0

    while loops < max_loops:
        loops += 1
        improved = False
        for block_len in block_sizes:
            for i in range(1, len(best_route) - 1 - block_len + 1):
                block = best_route[i:i + block_len]
                if instance.depot in block:
                    continue
                remaining = best_route[:i] + best_route[i + block_len:]

                for j in range(1, len(remaining)):
                    if j == i:
                        continue
                    new_route = remaining[:j] + block + remaining[j:]
                    if new_route[0] != instance.depot or new_route[-1] != instance.depot:
                        continue

                    new_cost = route_cost(new_route, instance)

                    if new_cost + 1e-9 < best_cost:
                        best_route = new_route
                        best_cost = new_cost
                        improved = True
                        break

                if improved:
                    break

            if improved:
                break

        if not improved:
            break

    return best_route

def improve_routes_oropt(routes, instance, block_sizes=(1, 2), max_loops=1):
    return [
        or_opt_route(route, instance, block_sizes=block_sizes, max_loops=max_loops)
        for route in routes
    ]

def extract_reference_value_from_vrp(file_path):
    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        text = f.read()

    patterns = [
        r"optimal value\s*:\s*([0-9]+(?:\.[0-9]+)?)",
        r"best value\s*:\s*([0-9]+(?:\.[0-9]+)?)",
    ]

    text_low = text.lower()
    for pattern in patterns:
        m = re.search(pattern, text_low)
        if m:
            return float(m.group(1))

    return None
```

```python
def parse_cvrplib_vrp(file_path: str) -> CVRPInstance:
    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        lines = [line.strip() for line in f if line.strip()]

    name = "unknown"
    dimension = None
    capacity = None
    depot = None

    coords = {}
    demands = {}
    edge_weights = []

    section = None
    edge_weight_format = None

    for line in lines:
        u = line.upper()

        if u.startswith("NAME"):
            name = line.split(":", 1)[1].strip()
        elif u.startswith("DIMENSION"):
            dimension = int(line.split(":", 1)[1].strip())
        elif u.startswith("CAPACITY"):
            capacity = int(line.split(":", 1)[1].strip())
        elif u.startswith("EDGE_WEIGHT_FORMAT"):
            edge_weight_format = line.split(":", 1)[1].strip()
        elif "NODE_COORD_SECTION" in u:
            section = "COORD"
        elif "DEMAND_SECTION" in u:
            section = "DEMAND"
        elif "EDGE_WEIGHT_SECTION" in u:
            section = "EDGE"
        elif "DEPOT_SECTION" in u:
            section = "DEPOT"
        elif u.startswith("EOF"):
            break
        else:
            parts = line.split()

            if section == "COORD" and len(parts) >= 3:
                node = int(parts[0])
                x = float(parts[1])
                y = float(parts[2])
                coords[node] = (x, y)

            elif section == "DEMAND" and len(parts) >= 2:
                node = int(parts[0])
                demand = int(parts[1])
                demands[node] = demand

            elif section == "DEPOT" and len(parts) >= 1:
                val = int(parts[0])
                if val != -1:
                    depot = val

            elif section == "EDGE":
                edge_weights.extend(map(float, parts))

    if dimension is None:
        raise ValueError(f"DIMENSION not found in {file_path}")
    if capacity is None:
        raise ValueError(f"CAPACITY not found in {file_path}")
    if depot is None:
        raise ValueError(f"DEPOT_SECTION not found in {file_path}")

    if coords:
        dist_matrix = build_distance_matrix(coords)
    else:
        if edge_weight_format != "LOWER_ROW":
            raise ValueError(f"Unsupported EDGE_WEIGHT_FORMAT in {file_path}: {edge_weight_format}")

        dist_matrix = np.zeros((dimension, dimension), dtype=float)
        idx = 0
        for i in range(1, dimension):
            for j in range(i):
                d = edge_weights[idx]
                dist_matrix[i, j] = d
                dist_matrix[j, i] = d
                idx += 1

        coords = {i + 1: (0.0, 0.0) for i in range(dimension)}

    return CVRPInstance(
        name=name,
        dimension=dimension,
        capacity=capacity,
        coords=coords,
        demands=demands,
        depot=depot,
        distance_matrix=dist_matrix
    )
```

```python
class ACOSolverCVRP:
    def __init__(
        self,
        instance: CVRPInstance,
        I: int = 200,
        q0: float = 0.85,
        beta: float = 4.0,
        alpha: float = 0.2,
        tau0: Optional[float] = None,
        m_ants: int = 20,
        patience: int = 50,
        candidate_k: int = 15,
        use_max_min: bool = True,
        tau_min_factor: float = 0.1,
        tau_max_factor: float = 5.0,
        global_weight: float = 2.0,
        seed: int = 42,
    ):
        self.instance = instance
        self.I = I
        self.q0 = q0
        self.beta = beta
        self.alpha = alpha
        self.m_ants = m_ants
        self.patience = patience
        self.candidate_k = candidate_k
        self.use_max_min = use_max_min
        self.tau_min_factor = tau_min_factor
        self.tau_max_factor = tau_max_factor
        self.global_weight = global_weight

        self.random = random.Random(seed)
        np.random.seed(seed)

        self.nodes = sorted(instance.coords.keys())
        self.n2i = node_to_index_map(instance)
        self.depot = instance.depot
        self.depot_idx = self.n2i[self.depot]
        self.num_nodes = len(self.nodes)

        self.eta = self._build_eta()
        self.candidate_list = self._build_candidate_list()

        if tau0 is None:
            tau0 = self._default_tau0()
        self.tau0 = tau0

        self.tau = np.full((self.num_nodes, self.num_nodes), self.tau0, dtype=float)
        np.fill_diagonal(self.tau, 0.0)

    def _build_eta(self):
        d = self.instance.distance_matrix.copy()
        eta = np.zeros_like(d, dtype=float)
        for i in range(d.shape[0]):
            for j in range(d.shape[1]):
                if i != j and d[i, j] > 0:
                    eta[i, j] = 1.0 / d[i, j]
        return eta

    def _default_tau0(self):
        d = self.instance.distance_matrix
        positive = d[d > 0]
        min_d = positive.min() if len(positive) > 0 else 1.0
        n = self.instance.dimension
        return 1.0 / ((n + 1) * min_d)

    def _feasible_candidates(self, current_node, unvisited, current_load):
        preferred = [
            node for node in self.candidate_list[current_node]
            if node in unvisited and current_load + self.instance.demands[node] <= self.instance.capacity
        ]

        if preferred:
            return preferred

        return [
            node for node in unvisited
            if current_load + self.instance.demands[node] <= self.instance.capacity
        ]

    def _choose_next_customer(self, current_node, candidates):
        i = self.n2i[current_node]

        desirabilities = []
        for node in candidates:
            j = self.n2i[node]
            value = self.tau[i, j] * (self.eta[i, j] ** self.beta)
            desirabilities.append(value)

        desirabilities = np.array(desirabilities, dtype=float)

        if desirabilities.sum() <= 0:
            return self.random.choice(candidates)

        if self.random.random() < self.q0:
            return candidates[int(np.argmax(desirabilities))]

        probs = desirabilities / desirabilities.sum()
        idx = np.random.choice(len(candidates), p=probs)
        return candidates[int(idx)]

    def _local_update(self, a, b):
        i = self.n2i[a]
        j = self.n2i[b]
        self.tau[i, j] = (1 - self.alpha) * self.tau[i, j] + self.alpha * self.tau0
        self.tau[j, i] = self.tau[i, j]

    def _global_evaporation(self):
        self.tau *= (1 - self.alpha)

    def _global_deposit(self, routes, cost):
        if cost <= 0:
            return
        deposit = self.alpha / cost
        for route in routes:
            for a, b in zip(route[:-1], route[1:]):
                i = self.n2i[a]
                j = self.n2i[b]
                self.tau[i, j] += deposit
                self.tau[j, i] += deposit

    def _additional_global_deposit(self, routes, cost):
        if cost <= 0:
            return
        deposit = self.global_weight * self.alpha / cost
        for route in routes:
            for a, b in zip(route[:-1], route[1:]):
                i = self.n2i[a]
                j = self.n2i[b]
                self.tau[i, j] += deposit
                self.tau[j, i] += deposit

    def _build_candidate_list(self):
        candidate_list = {}

        for node in self.nodes:
            i = self.n2i[node]

            neighbors = []
            for other in self.nodes:
                if other == node:
                    continue
                j = self.n2i[other]
                neighbors.append((other, self.instance.distance_matrix[i, j]))

            neighbors.sort(key=lambda x: x[1])
            candidate_list[node] = [node_id for node_id, _ in neighbors[:self.candidate_k]]

        return candidate_list

    def _clip_tau(self):
        if not self.use_max_min:
            return

        tau_min = self.tau0 * self.tau_min_factor
        tau_max = self.tau0 * self.tau_max_factor

        self.tau = np.clip(self.tau, tau_min, tau_max)
        np.fill_diagonal(self.tau, 0.0)

    def _construct_solution(self):
        unvisited = set(node for node in self.nodes if node != self.depot)
        routes = []

        while unvisited:
            route = [self.depot]
            current_node = self.depot
            current_load = 0

            while True:
                candidates = self._feasible_candidates(current_node, unvisited, current_load)

                if not candidates:
                    if route[-1] != self.depot:
                        route.append(self.depot)
                        self._local_update(current_node, self.depot)
                    break

                next_node = self._choose_next_customer(current_node, candidates)
                route.append(next_node)
                self._local_update(current_node, next_node)

                current_load += self.instance.demands[next_node]
                unvisited.remove(next_node)
                current_node = next_node

            if route[-1] != self.depot:
                route.append(self.depot)

            routes.append(route)

        total_cost = solution_cost(routes, self.instance)
        return routes, total_cost

    def solve(self, verbose=False):
        start = time.time()

        best_routes = None
        best_cost = float("inf")
        history = []
        no_improve = 0

        for iteration in range(1, self.I + 1):
            iteration_solutions = []
            for _ in range(self.m_ants):
                routes, cost = self._construct_solution()
                iteration_solutions.append((routes, cost))

            iter_best_routes, iter_best_cost = min(iteration_solutions, key=lambda x: x[1])
            iter_best_routes = improve_routes_2opt(iter_best_routes, self.instance)
            iter_best_routes = improve_routes_oropt(iter_best_routes, self.instance, block_sizes=(1, 2), max_loops=3)
            iter_best_routes = improve_routes_relocate(iter_best_routes, self.instance, max_loops=3)
            iter_best_routes = improve_routes_swap(iter_best_routes, self.instance, max_loops=3)
            iter_best_cost = solution_cost(iter_best_routes, self.instance)

            if iter_best_cost < best_cost:
                best_cost = iter_best_cost
                best_routes = iter_best_routes
                no_improve = 0
            else:
                no_improve += 1

            if no_improve >= self.patience:
                self.tau = (1 - self.alpha) * self.tau + self.alpha * self.tau0
                no_improve = 0

            self._global_evaporation()
            self._global_deposit(iter_best_routes, iter_best_cost)

            if best_routes is not None:
                self._global_deposit(best_routes, best_cost)
                self._additional_global_deposit(best_routes, best_cost)

            if hasattr(self, "_clip_tau"):
                self._clip_tau()

            history.append(best_cost)

            if verbose and (iteration == 1 or iteration % 20 == 0):
                print(f"Iter {iteration} | best={best_cost:.3f} | no_improve={no_improve}")

        elapsed = time.time() - start

        return ACOResult(
            best_cost=best_cost,
            best_routes=best_routes,
            history=history,
            elapsed_time=elapsed,
            iterations_done=len(history)
        )
```

```python
def infer_group_from_filename(filename: str) -> str:
    return filename[0].upper()


runs_per_instance = 3
all_rows = []
errors = []

for root, dirs, files in os.walk("."):
    for f in sorted(files):
        if not f.endswith(".vrp"):
            continue

        path = os.path.join(root, f)

        try:
            instance = parse_cvrplib_vrp(path)
            reference_value = extract_reference_value_from_vrp(path)
            data_group = infer_group_from_filename(f)

            for param_set in parameter_sets:
                run_costs = []
                run_gaps = []
                run_times = []

                for seed in range(runs_per_instance):
                    solver = ACOSolverCVRP(
                      instance=instance,
                      I=param_set["I"],
                      q0=param_set["q0"],
                      beta=param_set["beta"],
                      alpha=param_set["alpha"],
                      tau0=param_set["tau0"],
                      m_ants=param_set["m_ants"],
                      patience=param_set["patience"],
                      candidate_k=param_set["candidate_k"],
                      use_max_min=param_set["use_max_min"],
                      tau_min_factor=param_set["tau_min_factor"],
                      tau_max_factor=param_set["tau_max_factor"],
                      global_weight=param_set["global_weight"],
                      seed=seed,
                  )

                    result = solver.solve(verbose=False)

                    run_costs.append(result.best_cost)
                    run_times.append(result.elapsed_time)

                    if reference_value is not None:
                        gap = (result.best_cost - reference_value) / reference_value * 100
                        run_gaps.append(gap)

                row = {
                    "file": f,
                    "group": data_group,
                    "dimension": instance.dimension,
                    "set_name": param_set["set_name"],
                    "I": param_set["I"],
                    "q0": param_set["q0"],
                    "beta": param_set["beta"],
                    "alpha": param_set["alpha"],
                    "tau0": param_set["tau0"],
                    "m_ants": param_set["m_ants"],
                    "patience": param_set["patience"],
                    "candidate_k": param_set["candidate_k"],
                    "use_max_min": param_set["use_max_min"],
                    "tau_min_factor": param_set["tau_min_factor"],
                    "tau_max_factor": param_set["tau_max_factor"],
                    "global_weight": param_set["global_weight"],
                    "reference_value": reference_value,
                    "best_cost_min": float(np.min(run_costs)),
                    "best_cost_mean": float(np.mean(run_costs)),
                    "time_mean_sec": float(np.mean(run_times)),
                    "gap_mean_percent": float(np.mean(run_gaps)) if len(run_gaps) > 0 else np.nan,
                    "gap_best_percent": float(np.min(run_gaps)) if len(run_gaps) > 0 else np.nan,
                    "gap_std_percent": float(np.std(run_gaps)) if len(run_gaps) > 0 else np.nan,
                }

                all_rows.append(row)

                print(
                    f"{f:15s} | {param_set['set_name']:18s} | "
                    f"cost_mean={row['best_cost_mean']:.3f} | "
                    f"gap_mean={row['gap_mean_percent'] if not np.isnan(row['gap_mean_percent']) else 'N/A'} | "
                    f"time_mean={row['time_mean_sec']:.3f}s"
                )

        except Exception as e:
            errors.append((path, str(e)))
            print("ERROR:", path, e)

results_df = pd.DataFrame(all_rows)
```
